```python
df["HAS_MACRO_CELL"] = df["MACRO CELL"].eq("YES")
df["HAS_SMALL_CELL"] = df["SMALL CELL"].eq("YES")
df["HAS_PROXIMITY_CELL"] = df["PROXIMITY TO CELL"].eq("YES")
```

```python
df[
    [
        "SITE NAME",
        "HAS_MACRO_CELL",
        "HAS_SMALL_CELL",
        "HAS_PROXIMITY_CELL"
    ]
].head()
```

```python
df["COVERAGE_COUNT"] = (
    df[
        [
            "HAS_MACRO_CELL",
            "HAS_SMALL_CELL",
            "HAS_PROXIMITY_CELL"
        ]
    ]
    .sum(axis=1)
)
```

```python
df["COVERAGE_COUNT"].value_counts()
```

```python
def get_coverage_type(row):
    types = []

    if row["HAS_MACRO_CELL"]:
        types.append("Macro Cell")

    if row["HAS_SMALL_CELL"]:
        types.append("Small Cell")

    if row["HAS_PROXIMITY_CELL"]:
        types.append("Proximity to Cell")

    return " + ".join(types)


df["COVERAGE_TYPE"] = df.apply(get_coverage_type, axis=1)
```

```python
pop_check = df.groupby("SITE TYPE")["POPULATION"].agg(
    total_rows="size",
    known_population="count"
)

pop_check["missing_population"] = (
    pop_check["total_rows"] - pop_check["known_population"]
)

pop_check["missing_percent"] = (
    pop_check["missing_population"]
    / pop_check["total_rows"]
    * 100
).round(1)

pop_check.sort_values("missing_percent", ascending=False)
```
